```python
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import LabelEncoder, StandardScaler, PolynomialFeatures
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score, mean_absolute_percentage_error
from sklearn.decomposition import PCA
from sklearn.cross_decomposition import PLSRegression
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
```

```python
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
categorical_cols = df.select_dtypes(include=['object', 'str']).columns.tolist()

print(numerical_cols)
print(categorical_cols)
```

```python
df = pd.get_dummies(df, columns=categorical_cols, drop_first=True)
```

```python
df.shape
```

```python
x = df.drop(columns=["SalePrice"])
y = df["SalePrice"]
```

```python
for col in x.columns:
    if (col == "SalePrice"):
        print("Drop Failed")
        break
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, random_state=434)
sc = StandardScaler()
x_test = sc.fit_transform(x_test)
x_train = sc.fit_transform(x_train)
```

```python
def create_metrics(mae, rmse, mape, r2):
    adjr2 = 1 - (1 - r2) * (x.shape[0] - 1) / (x.shape[0] - x.shape[1] - 1)

    data = {
        "Metrics": ["Mean Absolute Error", "Root Mean Squared Error", "Mean Absolute Percentage Error", "R Squared", "Adjusted R Squared"],
        "Values": [round(mae, 2), round(rmse, 2), round(mape, 2), round(r2, 2), round(adjr2, 2)],
        "Explanation" : ["Pure error", "Penalizes big misses", "More intepretable MAE", "Overall fit", "Useful for subset selecting models"]  
            }
    
    metrics = pd.DataFrame(data)

    return metrics
```

```python
model = LinearRegression()
model.fit(x_train, y_train)
y_pred = model.predict(x_test)
mae = mean_absolute_error(y_test, y_pred)
mape = mean_absolute_percentage_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

metrics = create_metrics(mae, rmse, mape, r2)
print("Linear Regression:")
metrics.head()
```

```python
pipeline2 = Pipeline([
    ('poly', PolynomialFeatures(degree = 2)),
    ('model', LinearRegression())
])

pipeline2.fit(x_train, y_train)

y_pred = pipeline2.predict(x_test)
mae = mean_absolute_error(y_test, y_pred)
mape = mean_absolute_percentage_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

metrics = create_metrics(mae, rmse, mape, r2)
print("Polynomial Regression (Degree = 2):")
metrics.head()
```

```python
pipeline3 = Pipeline([
    ('poly', PolynomialFeatures(degree = 3)),
    ('model', LinearRegression())
])

pipeline3.fit(x_train, y_train)

y_pred = pipeline3.predict(x_test)
mae = mean_absolute_error(y_test, y_pred)
mape = mean_absolute_percentage_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

metrics = create_metrics(mae, rmse, mape, r2)
print("Polynomial Regression (Degree = 3):")
metrics.head()
```

```python
from sklearn.linear_model import LassoCV

model = LassoCV(cv=5, random_state=434, max_iter=10000).fit(x_train, y_train)

y_pred = model.predict(x_test)
mae = mean_absolute_error(y_test, y_pred)
mape = mean_absolute_percentage_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

metrics = create_metrics(mae, rmse, mape, r2)
print(f"Lasso Regression with an regularization parameter of {model.alpha_:.2f}")
metrics.head()
```

```python
ctr = 0
total = 0
dropped = []
for coef in model.coef_:
    if coef == 0:
        total = total + 1
        print(f"Feature {ctr} ({x.columns[ctr]}) set to 0")
        dropped.append(x.columns[ctr])
    ctr = ctr + 1

print(f"A total of {total} features were set to 0")
```

```python
scoring = {
    "MAE": 'neg_mean_absolute_error',
    "MAPE": 'neg_mean_absolute_percentage_error',
    "MSE": 'neg_mean_squared_error',
    "R2": 'r2'
}

params = {
    "n_estimators": (100, 200, 300, 400),
    "max_depth": (None, 10, 20),
    "max_features": ("sqrt", "log2", None),
}

model = RandomForestRegressor()

grid = GridSearchCV(estimator=model, param_grid=params, scoring=scoring, refit="MAE")

grid.fit(x_train, y_train)

y_pred = grid.predict(x_test)
mae = mean_absolute_error(y_test, y_pred)
mape = mean_absolute_percentage_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

metrics = create_metrics(mae, rmse, mape, r2)
print(f"Random Forests with the final parameters of: ")

for key, value in grid.best_params_.items():
    print(f"{key}: {value}")
metrics.head()
```

```python
import xgboost as xgb

scoring = {
    "MAE": 'neg_mean_absolute_error',
    "MAPE": 'neg_mean_absolute_percentage_error',
    "MSE": 'neg_mean_squared_error',
    "R2": 'r2'
}

params = {
    "n_estimators": (100, 200, 300, 400),
    "max_depth": (3, 6, 9),
    "learning_rate": (0.01, 0.1, 0.5, 1),
    "subsample": (0.33, 0.67, 1)
}

model = xgb.XGBRegressor(tree_method="hist", device="cuda")

grid = GridSearchCV(estimator=model, param_grid=params, scoring=scoring, refit="MAE")

grid.fit(x_train, y_train)

y_pred = grid.predict(x_test)
mae = mean_absolute_error(y_test, y_pred)
mape = mean_absolute_percentage_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

metrics = create_metrics(mae, rmse, mape, r2)
print(f"XGBoost with the final parameters of: ")

for key, value in grid.best_params_.items():
    print(f"{key}: {value}")
metrics.head()
```
